3. Why must we specify a value of s?
4. Estimate ζ(2) by adding its first four terms

```python
# 1. Write the first five terms of ζ(3)
display(
    Math(r"1.\; \displaystyle \zeta(3) = 1 + \frac{1}{2^3} + \frac{1}{3^3} + \frac{1}{4^3} + \frac{1}{5^3} + \dots"))

# 2. Explain what ζ(2) represents
zeta2 = sp.zeta
display(
    Math(
        r"\begin{aligned}"
            r"2.\; \displaystyle \zeta(2) &= \sum_{n=1}^{\infty} \frac{1}{n^2} \\"
            r"\displaystyle \zeta(2) &= " rf"{sp.latex(zeta2(2))}"
        r"\end{aligned}"
    )
)

# 3. Why must we specify a value of s?
display(Markdown(r"3. The Riemann zeta function $\zeta(s)$ is defined as a series $\sum_{n=1}^{\infty} \frac{1}{n^s}$, which depends on the value of $s$. Different values of $s$ give different series and thus different results, so it is necessary to specify $s$ to know which particular value of the zeta function we are referring to."))

# 4. Estimate ζ(2) by adding its first four terms
first_four_terms = sum(sp.Rational(1, k**2) for k in range(1, 5))

display(
    Math(
        rf"\text{{4. }} \zeta(2) \approx "
        rf"\sum_{{n=1}}^4 \frac{{1}}{{n^2}} = "
        rf"{sp.latex(first_four_terms)} \approx {first_four_terms.evalf()}"
    )
)
```
